```python
from pyspark.sql import SparkSession
```

```python
warehouse_location='/user/insider'
```

```python
spark =SparkSession \
    .builder \
    .appName("Mi primer aplicacion de Spark") \
    .config("spark.sql.warehouse.dir", warehouse_location) \
    .enableHiveSupport() \
    .getOrCreate()
```

```python
emails=spark.sql("select * from EMAIL_DATA limit 100")
```

```python
emails_df=emails.toPandas()
```
